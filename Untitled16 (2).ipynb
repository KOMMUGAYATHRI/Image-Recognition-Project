```python
import os
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf

from tensorflow.keras.preprocessing.image import ImageDataGenerator
```

```python
dataset_path = "/content/drive/MyDrive/FOOD_DATASET"

print("Path exists:", os.path.exists(dataset_path))
```

```python
print("Food Categories:")
print(os.listdir(dataset_path))
```

```python
for folder in os.listdir(dataset_path):

    folder_path = os.path.join(dataset_path, folder)

    if os.path.isdir(folder_path):

        images = [
            file for file in os.listdir(folder_path)
            if file.lower().endswith(
                ('.jpg', '.jpeg', '.png', '.webp')
            )
        ]

        print(folder, ":", len(images), "images")
```

```python
IMG_SIZE = 224
BATCH_SIZE = 8
```

```python
IMG_SIZE = (224, 224)
BATCH_SIZE = 4
SEED = 42

train_data = tf.keras.utils.image_dataset_from_directory(
    dataset_path,
    validation_split=0.2,
    subset="training",
    seed=SEED,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE
)

validation_data = tf.keras.utils.image_dataset_from_directory(
    dataset_path,
    validation_split=0.2,
    subset="validation",
    seed=SEED,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE
)

class_names = train_data.class_names

print("Classes:", class_names)
```

```python
data_augmentation = tf.keras.Sequential([
    tf.keras.layers.RandomFlip("horizontal"),
    tf.keras.layers.RandomRotation(0.1),
    tf.keras.layers.RandomZoom(0.1)
])

AUTOTUNE = tf.data.AUTOTUNE

train_data = train_data.map(
    lambda x, y: (data_augmentation(x, training=True), y),
    num_parallel_calls=AUTOTUNE
).prefetch(AUTOTUNE)

validation_data = validation_data.prefetch(AUTOTUNE)
```

```python
base_model = tf.keras.applications.MobileNetV2(
    input_shape=(224, 224, 3),
    include_top=False,
    weights="imagenet"
)

base_model.trainable = False

model = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(224, 224, 3)),

    tf.keras.layers.Lambda(
        tf.keras.applications.mobilenet_v2.preprocess_input
    ),

    base_model,

    tf.keras.layers.GlobalAveragePooling2D(),

    tf.keras.layers.Dropout(0.3),

    tf.keras.layers.Dense(
        len(class_names),
        activation="softmax"
    )
])
```

```python
model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)
```

```python
history = model.fit(
    train_data,
    validation_data=validation_data,
    epochs=10
)
```

```python
plt.plot(history.history['accuracy'])

plt.plot(history.history['val_accuracy'])

plt.title('Model Accuracy')

plt.xlabel('Epoch')

plt.ylabel('Accuracy')

plt.legend(['Training Accuracy', 'Validation Accuracy'])

plt.show()
```

```python
plt.plot(history.history['loss'])

plt.plot(history.history['val_loss'])

plt.title('Model Loss')

plt.xlabel('Epoch')

plt.ylabel('Loss')

plt.legend(['Training Loss', 'Validation Loss'])

plt.show()
```

```python
model.save(
    "/content/drive/MyDrive/food_recognition_model.keras"
)

print("Model saved successfully!")
```

```python
test_image_path = "/content/drive/MyDrive/test_food/images.jpg"
```

```python
from tensorflow.keras.preprocessing import image
import numpy as np
import matplotlib.pyplot as plt

img = image.load_img(
    test_image_path,
    target_size=(224, 224)
)

img_array = image.img_to_array(img)
img_array = np.expand_dims(img_array, axis=0)

prediction = model.predict(img_array)

predicted_index = np.argmax(prediction)

predicted_class = class_names[predicted_index]

confidence = np.max(prediction) * 100

plt.imshow(img)
plt.axis("off")
plt.title(
    f"Prediction: {predicted_class}\nConfidence: {confidence:.2f}%"
)
plt.show()

print("Predicted Food Category:", predicted_class)
print("Confidence:", f"{confidence:.2f}%")
```
